```python
# Missing Values
missing_values=df.isnull().sum()
missing_values=missing_values[missing_values > 0].sort_values(ascending=False)
print("Columns containing missing values:\n")
print(missing_values)
```

```python
# Missing Value Percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_table = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_table = missing_table[
    missing_table["Missing Count"] > 0].sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_table
```

```python
# Duplicate Records
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
# Unique Values
unique_values = df.nunique().sort_values()
unique_values
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"],errors="coerce")
print(df["Order_Date"].dtype)
```

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()
df["Quarter"] = df["Order_Date"].dt.quarter

print(df[[
    "Order_Date",
    "Year",
    "Month",
    "Month_Name",
    "Quarter"
]].head())
```

```python
df.describe()
```

```python
# Finding Mean
numerical_columns = df.select_dtypes(include=np.number).columns
mean_values = df[numerical_columns].mean()
print("Mean:\n")
print(mean_values)
```

```python
# Finding Median
median_values = df[numerical_columns].median()
print("Median:\n")
print(median_values)
```

```python
# Finding Mode
mode_values = df[numerical_columns].mode().iloc[0]
print("Mode:\n")
print(mode_values)
```

```python
# Finding Standard Deviation
std_values = df[numerical_columns].std()
print("Standard Deviation:\n")
print(std_values)
```

```python
# Statistics of Mean, Median, Mode
statistics_summary = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Mode": df[numerical_columns].mode().iloc[0],
    "Standard Deviation": df[numerical_columns].std()
})

statistics_summary
```

```python
# Total Sales
total_sales = df["Sales_Amount"].sum()
print(f"Total Sales Revenue: ${total_sales:,.2f}")
```

```python
# Total Profit
total_profit = df["Profit"].sum()
print(f"Total Profit: ${total_profit:,.2f}")
```

```python
# Total Quantity
total_quantity = df["Quantity"].sum()
print("Total Quantity Sold:", total_quantity)
```

```python
# Average Order vales
average_order_value = df["Sales_Amount"].mean()
print(f"Average Transaction Value: ${average_order_value:,.2f}")
```

```python
# Profit Margin
profit_margin = (df["Profit"].sum() / df["Sales_Amount"].sum()) * 100
print(f"Overall Profit Margin: {profit_margin:.2f}%")
```

```python
# Monthly Sales
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Sales_Amount"].sum()
)
monthly_sales.index = monthly_sales.index.astype(str)
monthly_sales.head()
```

```python
# Monthly Sales Plot chart
plt.figure(figsize=(16, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### Observation

The monthly sales trend shows how revenue changes over time. Peaks indicate periods of stronger sales activity, while lower points represent relatively weaker sales months. These fluctuations can help the business identify seasonal patterns and plan inventory and marketing activities accordingly.

```python
quarterly_sales = (
    df.groupby(
        [df["Order_Date"].dt.year,
         df["Order_Date"].dt.quarter]
    )["Sales_Amount"]
    .sum()
)
quarterly_sales
```

```python
quarterly_sales = (
    df.groupby(
        [df["Order_Date"].dt.year,
         df["Order_Date"].dt.quarter]
    )["Sales_Amount"].sum()
)

quarterly_sales
```

```python
quarterly_sales.index.names = ['Year', 'Quarter']
quarterly_sales_plot = quarterly_sales.reset_index()

quarterly_sales_plot["Quarter_Label"] = (
    quarterly_sales_plot["Year"].astype(str)
    + "-Q"
    + quarterly_sales_plot["Quarter"].astype(str)
)

print(quarterly_sales_plot.head())
```

```python
quarterly_sales = (
    df.groupby(
        df["Order_Date"].dt.to_period("Q")
    )["Sales_Amount"]
    .sum()
)

quarterly_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker="o"
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Sales Amount")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

Quarterly sales provide a higher-level view of business performance and make it easier to identify seasonal patterns. Strong quarters can be used as benchmarks for future sales planning, while weaker quarters may require additional promotional campaigns.

```python
# Customer Age Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Customer_Age",
    bins=20,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.show()
```

### Observation

The age distribution shows the age groups contributing most frequently to transactions. This information can help the business understand its dominant customer demographic and design targeted marketing campaigns.

```python
# Create Age Groups
bins = [0, 18, 25, 35, 45, 55, 65, 100]
labels = ["Under 18","18-25","26-35","36-45","46-55","56-65","65+"]
df["Age_Group"] = pd.cut(
    df["Customer_Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
df["Age_Group"].value_counts().sort_index()
```

```python
# Age Group Distribution
age_group_counts = (
    df["Age_Group"]
    .value_counts()
    .sort_index()
)
plt.figure(figsize=(10, 6))
sns.barplot(
    x=age_group_counts.index,
    y=age_group_counts.values
)
plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The age-group analysis identifies which customer age ranges contribute the most transactions. These segments can be targeted with age-appropriate products, promotions, and communication strategies.

```python
# Gender Breakdown
gender_counts = df["Customer_Gender"].value_counts()
gender_counts
```

```python
plt.figure(figsize=(8, 6))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
)

plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")

plt.show()
```

### Observation

The gender distribution shows the relative contribution of different customer-reported gender groups. This can help the business evaluate whether its product and marketing strategies are reaching its customer base evenly.

```python
# Top 10 Best-Selling Products
top_products = (
    df.groupby("Product_Name")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
top_products
```

### Observation

The top-selling products represent the products with the highest unit demand. These products should receive appropriate inventory priority to reduce the risk of stockouts and maintain customer satisfaction.

```python
# Best Performing Categories
category_summary = (
    df.groupby("Product_Category")
    .agg(
        Total_Sales=("Sales_Amount", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Average_Sales=("Sales_Amount", "mean")
    )
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

category_summary
```

```python
# Overall Business KPI Table
kpi = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Total Quantity Sold",
        "Total Revenue",
        "Total Profit",
        "Average Transaction Value",
        "Profit Margin",
        "Unique Customers",
        "Unique Products"
    ],

    "Value": [
        len(df),
        df["Quantity"].sum(),
        df["Sales_Amount"].sum(),
        df["Profit"].sum(),
        df["Sales_Amount"].mean(),
        profit_margin,
        df["Customer_ID"].nunique(),
        df["Product_ID"].nunique()
    ]
})

kpi
```

# Key Business Insights

Based on the exploratory data analysis, the following insights were identified:

1. **Sales Trend:** Monthly and quarterly analysis reveals changes in sales performance over time and helps identify strong and weak periods.

2. **Customer Demographics:** Customer age-group and gender analysis identifies the major demographic groups contributing to transactions.

3. **Product Performance:** The top-selling product analysis identifies products with the highest demand and helps prioritize inventory.

4. **Category Performance:** Revenue and profit by product category identify the strongest categories for business focus.

5. **Sales Channels:** Comparing sales channels reveals which channels contribute the greatest amount of revenue.

6. **Geographical Performance:** Regional and city-level analysis highlights areas with stronger sales performance.

7. **Profitability:** Comparing sales and profit helps identify whether high-revenue transactions are also generating strong profitability.

8. **Customer Behaviour:** Customer segments, ratings, and transaction characteristics provide additional information about customer behaviour.

# Actionable Business Recommendations

### 1. Optimize Inventory for High-Demand Products

The business should maintain sufficient inventory for the top-selling products identified during the analysis. These products generate significant transaction volume, so avoiding stockouts can help maintain sales and customer satisfaction.

### 2. Focus Marketing on High-Performing Customer Segments

Customer demographic and segment analysis should be used to create targeted marketing campaigns. Promotions can be customized based on the age groups, customer segments, and sales channels that contribute the most revenue.

### 3. Strengthen High-Performing Product Categories

The categories generating the highest revenue and profit should receive greater marketing and inventory attention. Lower-performing categories should be reviewed for pricing, promotions, product assortment, and customer demand before additional resources are allocated.
